```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('wild_rift_fake_data.csv')
```

```python
df['herois'][28]
```

```python
df.drop(index=28, inplace=True)
```

```python
df['herois']
```

```python
df.head()
```

```python
hero_mapping = {hero: i for i, hero in enumerate(df['herois'].unique())}
df['herois'] = df['herois'].map(hero_mapping)
df['herois']
```

```python
df.head()
```

```python
df.info()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop('herois', axis=1), df['herois'], test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model=LinearRegression()
```

```python
model.fit(X_train, y_train)
```

```python
predictions = model.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
```

```python
mse = mean_squared_error(y_test, predictions)
r2 = r2_score(y_test, predictions)
mae = mean_absolute_error(y_test, predictions)
```

```python
print(mse)
print(r2)
print(mae)
```

```python
my_stats = pd.DataFrame({
    'Win_rate': [70.0],
    'DPS': [7000],
    'Assistance': [5],
    'Kills': [10]
})

reverse_hero_mapping = {v: k for k, v in hero_mapping.items()}

My_Champion = model.predict(my_stats)

My_Champion = int(My_Champion[0])

My_Champion_Name = reverse_hero_mapping[My_Champion]

print('The Champion you should pick is: ', My_Champion_Name)
```
