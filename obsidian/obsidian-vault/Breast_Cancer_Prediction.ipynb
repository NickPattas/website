```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv('breast cancer.csv')
```

```python
df.head()
```

```python
df.drop(['Unnamed: 32', 'id'],axis=1,inplace=True)
```

```python
df.head()
```

```python
df.info()
```

```python
df['diagnosis'].unique()
```

```python
df['diagnosis']=df['diagnosis'].map({'M':1,'B':0})
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df['diagnosis'].unique()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop('diagnosis',axis=1), df['diagnosis'], test_size=0.3, random_state=101)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(max_iter=5000)
model.fit(X_train,y_train)
```

```python
predictions = model.predict(X_test)
print(predictions)
```

```python
from sklearn.metrics import classification_report
```

```python
classification_report(y_test,predictions)
```

```python
random_patient_data = {}
for col in X_train.columns:
    min_val = X_train[col].min()
    max_val = X_train[col].max()
    random_patient_data[col] = [np.random.uniform(min_val, max_val)]

random_patient_df = pd.DataFrame(random_patient_data)

prediction = model.predict(random_patient_df)
if prediction[0] == 1:
    print("Predicted diagnosis: Malignant")
else:
    print("Predicted diagnosis: Benign")
```
